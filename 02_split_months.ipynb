```python
import pandas as pd
df_2009_2010 = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2009-2010")
df_2010_2011 = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")
df = pd.concat([df_2009_2010, df_2010_2011], ignore_index=True)
df.shape
```

```python
df['year_month'] = df['InvoiceDate'].dt.to_period('M')
df['year_month'].unique()
```

```python
import os
os.makedirs("monthly_files", exist_ok=True)
```

```python
for period in df['year_month'].unique():
    month_data = df[df['year_month'] == period]
    filename = f"monthly_files/{period}.xlsx"
    month_data.to_excel(filename, index=False)
    print(f"Saved {filename} with {len(month_data)} rows")
```
