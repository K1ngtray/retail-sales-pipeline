```python
import pandas as pd
import os

files = os.listdir("monthly_files")
files
```

```python
all_months = []

for filename in files:
    filepath = os.path.join("monthly_files", filename)
    month_df = pd.read_excel(filepath)
    all_months.append(month_df)

combined = pd.concat(all_months, ignore_index=True)
combined.shape
```

```python
combined.columns
```

```python
combined = combined.drop(columns=['year_month'])
combined.shape
```

```python
combined.to_csv("combined_raw.csv", index=False)
```
