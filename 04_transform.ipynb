```python
import pandas as pd

df = pd.read_csv("combined_raw.csv")
df.shape
```

```python
df.dtypes
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df.dtypes
```

```python
non_product_codes = ['POST', 'M', 'DOT', 'BANK CHARGES']

non_products = df[df['StockCode'].isin(non_product_codes)]
products_only = df[~df['StockCode'].isin(non_product_codes)]

print(f"Non-product rows: {len(non_products)}")
print(f"Product rows: {len(products_only)}")
```

```python
non_products.to_csv("non_product_rows.csv", index=False)
```

```python
df = products_only
df.shape
```

```python
df['is_cancellation'] = df['Invoice'].astype(str).str.startswith('C')
df['is_cancellation'].value_counts()
```

```python
df['line_total'] = df['Quantity'] * df['Price']
df[['Quantity', 'Price', 'line_total']].head()
```
